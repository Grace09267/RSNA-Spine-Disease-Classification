```python
import pandas as pd

# 파일 불러오기
df = pd.read_csv("slice_level_labeled.csv")

# 라벨 컬럼만 추출 (study_id, series_id, instance_number 제외)
label_cols = df.columns.difference(["study_id", "series_id", "instance_number"])

# 각 행에 대해, 하나라도 유효한 라벨(-1이 아님)이 있으면 True
valid_mask = (df[label_cols] != -1).any(axis=1)

# 필터링
filtered_df = df[valid_mask].reset_index(drop=True)

# 저장
filtered_df.to_csv("slice_level_labeled_filtered.csv", index=False)
print(f"✅ 유효한 슬라이스 수: {len(filtered_df)}개 저장 완료")
```

```python
import os
import pandas as pd
import numpy as np
from tqdm import tqdm

# 기존 필터링한 라벨 포함된 DataFrame
df = pd.read_csv("slice_level_labeled_filtered.csv")
label_cols = df.columns.difference(["study_id", "series_id", "instance_number"])

final_rows = []

for i, row in tqdm(df.iterrows(), total=len(df)):
    study_id = int(row["study_id"])
    series_id = int(row["series_id"])
    instance_number = int(row["instance_number"])
    
    # 좌표 파일 존재 확인
    coord_path = f"heatmaps/{study_id}_{series_id}_{instance_number}.npy"
    if not os.path.exists(coord_path):
        continue
    
    # 라벨이 하나라도 유효한 경우만 통과
    labels = row[label_cols].values
    if np.any(labels != -1):
        final_rows.append(row)

# 결과 저장
final_df = pd.DataFrame(final_rows)
final_df.to_csv("slice_level_labeled_final.csv", index=False)
print(f"✅ 최종 필터링된 유효 슬라이스 수: {len(final_df)}")
```
